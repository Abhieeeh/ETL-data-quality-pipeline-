```python
def error_count(path,dataset):
    df=pd.read_csv(path)
    error=(
        df["validation_error"]
        .str.split(";")
        .explode()
        .str.strip())
    error=pd.DataFrame(error)
    print(error.shape)
    error=error[error!=""]
    count=error.groupby("validation_error").size().values
    index=error.groupby("validation_error").size().index
    error_details=pd.DataFrame({
    "Dataset":dataset,
    "Error":index,
    "count":count
    })
    # 
    
```
